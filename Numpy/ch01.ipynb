## 01. The NumPy ndarray: A Multidimensional Array Object

In numpy we can create N-dimensional arrays or ndarrays which is a fast, flexible, container for large datasets in python. Example

```python
data =  np.array([[2,4,5,6], [20,30,50,60]])
print(data)

#printing the number of dimensiona for above array
print(f"Number of Dimensions: {np.ndim(data)}")

#for the shape we can basically think of it as row and columns
print(f"Array shape: {data.shape}")
```

*dtype* represents the datatype for that array. It is important to ensure that all data in the array is of the same datatype. For example if we have an array made up of 10 integer values and then there is 1 float value, then the entire array is of data type float.

```python
data.dtype
```

### 01. Creating ndarrays


In additional to np.array, there are a number of other functions for creating new arrays, As examples, *np.zeros* and *np.ones* create arrays of 0s or 1s, respectively with a given length or shape. *np.empty* creates an array without initializing its values to any particular values. 

```python
np.zeros(10) 
```

```python
np.zeros((3,5)) # zeros array with shape 3x5
```

```python
np.ones(5)
```

```python
np.empty((2,3,2))
```

np.arange is an array value version of the built in Python range function

```python
np.arange(50)
```

### 02. Data Types for ndarrays

The data type or dtype is a special object containing the information (or metadata, data about the data) the ndarray needs to interpret a chunk of memory as a particular type of data

```python
arr1 = np.array([1,2,3], dtype=np.float64)
arr1.dtype
```

```python
arr2 = np.array([1,2,3], dtype=np.int32)
arr2.dtype
```

We can also explicitly convert or cast an array from one data type to another using ndarray's **astype** methon

```python
arr3 =  np.array([-1.2,4.2,-0.2,6.5])
arr3.dtype
```

```python
#Castign arr3 from float to inter
arr3.astype("int32")
```

If you have an array of strings representing numbers, you can use astype to convert
them to numeric form:


```python
numeric_strings = np.array(["1.25", "-9.6", "42"], dtype=np.bytes_)
numeric_strings.dtype
```

```python
numeric_strings.astype(float)
```
