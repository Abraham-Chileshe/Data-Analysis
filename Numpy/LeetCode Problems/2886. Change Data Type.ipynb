# 2886. Change Data Type

```python
import pandas as pd

students = pd.DataFrame({
    "student_id": [1, 2],
    "name": ["Ava", "Kate"],
    "age": [6, 15],
    "grade": [73.0, 87.0]
})

students["grade"] = students["grade"].astype(int)
students
```

**Leetcode Solution**

```python
import pandas as pd

def changeDatatype(students: pd.DataFrame) -> pd.DataFrame:
    students["grade"] = students["grade"].astype(int)
    return students
```
